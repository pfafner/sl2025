```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from mpl_toolkits.mplot3d import Axes3D

from sklearn.datasets import make_circles
from sklearn.model_selection import train_test_split

from sklearn.decomposition import PCA, KernelPCA
```

## 1. Generar datos

```python
seed = 2023

X, y = make_circles(n_samples=1_000, factor=0.3, noise=0.05, random_state=seed)
X_train, X_test, y_train, y_test = train_test_split(X, y, stratify=y, random_state=seed)
```

```python
X
```

```python
y[:100]
```

```python
plt.figure(figsize=(8,4))

plt.subplot(1,2,1)
plt.scatter(X_train[:,0], X_train[:,1], c=y_train)
plt.title('Train data')
plt.subplot(1,2,2)
plt.scatter(X_test[:,0], X_test[:,1], c=y_test)
plt.title('Test data')

plt.show()
```

### Polynomial Kernel

```python
k = 2

pca = PCA(n_components=k)
kernel_pca = KernelPCA(n_components=None, kernel="poly", degree=3, gamma=5, fit_inverse_transform=True, alpha=0.1)
```

```python
# Transform data
X_test_pca = pca.fit(X_train).transform(X_test)
X_test_kernel_pca = kernel_pca.fit(X_train).transform(X_test)
```

```python
plt.figure(figsize=(8,4))

plt.subplot(1,2,1)
plt.scatter(X_test_pca[:,0], X_test_pca[:,1], c=y_test)
plt.title('PCA')
plt.subplot(1,2,2)
plt.scatter(X_test_kernel_pca[:,0], X_test_kernel_pca[:,1], c=y_test)
plt.title('Kernel PCA')
plt.show()
```
